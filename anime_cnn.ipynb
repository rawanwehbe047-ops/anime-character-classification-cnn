## 3. How to understand the results
- **Training loss:** how much error the model makes on the training data; lower is generally better.
- **Validation loss:** error on images held out during training.
- **Validation accuracy:** the percentage of validation images classified correctly.

The dataset is small, so accuracy can change between runs and does not prove the model will work well on new images. This is a learning project, not a production-ready classifier.
